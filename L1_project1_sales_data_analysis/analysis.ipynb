```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("sales_data_sample.csv", encoding="latin1")
df.head()
df.info()
df.describe()
df.dropna()
df.drop_duplicates()
df["ORDERDATE"] = pd.to_datetime(df["ORDERDATE"])
total_sales=df["SALES"].sum()
print("\n Total Sales : ",total_sales)
avg_sales=df["SALES"].mean()
print("\n Average Sales : ",avg_sales)
max_sales=df["SALES"].max()
print("\n Maximum Sales : ",max_sales)
min_sales=df["SALES"].min()
print("\n Minimum Sales : ",min_sales)
category_sales=df.groupby("PRODUCTLINE")["SALES"].sum()
print("\n Category Sales : ",category_sales)
region_sales=df.groupby("COUNTRY")["SALES"].sum()
print("\n Country Sales : ",region_sales)
```

```python
df.groupby("PRODUCTLINE")["SALES"].sum().plot(kind='bar')
```

```python
df.groupby("COUNTRY")["SALES"].sum().plot(kind='pie')
```

```python
df.resample("ME", on="ORDERDATE")["SALES"].sum().plot()
```

```python
df.groupby("PRODUCTLINE")["SALES"].sum().nlargest(5).plot(kind='bar')
```
